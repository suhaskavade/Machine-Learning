# IPL Matches Seaborn Distribution

```python
import seaborn as sns
import pandas as pd
```

```python
df = pd.read_csv('/content/class_ipl_22-26_dataset.csv')
df.tail(74)
```

# Relational Plot

```python
sns.relplot(
    data=df.tail(74),
    x='team1_runs',
    y='team2_runs',
    kind='scatter'
)
```

# Scatter Plot: Team 1 Runs vs Team 2 Runs

```python
sns.scatterplot(
    x='team1_runs',
    y='team2_runs',
    hue='winner',
    data=df.tail(74)
)
```

# Scatter Plot: Team 1 Runs vs Team 1 Wickets

```python
sns.scatterplot(
    x='team1_runs',
    y='team1_wickets',
    data=df.tail(74)
)
```

# Count Plot

```python
sns.countplot(
    x='winner',
    data=df.tail(74),
).tick_params(axis='x',rotation=90)
```

# Bar Plot : Season vs Team 1 Runs

```python
sns.barplot(
    x='season',
    y='team1_runs',
    data = df.tail(74),
    #errorbar=None
).tick_params(axis='x',rotation=90)
```

# Bar Plot : Season vs Win By Wickets

```python
sns.barplot(
    x='team1',
    y='team1_runs',
    data = df.tail(74)
).tick_params(axis='x',rotation=90)
```

# Distribution Plot : Team 2 Runs

```python
sns.displot(
    df['team2_runs'].tail(74),
    kde=True
)
```

# Histogram

```python
sns.histplot(
    df['winner'],
    kde=True
).tick_params(axis='x',rotation=90)
```

# Boxplot: Team 1 Runs

```python
sns.boxplot(
    x=df['team1_runs'].tail(74)
)
```

# Boxplot : Team 2 Runs

```python
sns.boxplot(
    x=df['team2_runs'].tail(74)
)
```

# Pair Plot

```python
sns.pairplot(
    df[['team1_runs','team2_runs']].tail(74)
)
```

# Correlation matrix

```python
correlation = df.corr(numeric_only=True)
correlation
```

# Heatmap

```python
sns.heatmap(
    correlation,
    annot=True,
    cmap='Blues'
)
```

# KDE Plot

```python
sns.kdeplot(
    df['team1_runs'].tail(74),
    fill=True,
    bw_adjust=0.5
)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
df = pd.read_csv('/content/class_ipl_22-26_dataset.csv')
fig, axes = plt.subplots(3, 2, figsize=(16,18))

# 1. Winner Count Plot
winner = df['winner'].value_counts().head().index

sns.countplot(
    x='winner',
    data=df,
    order=winner,
    ax=axes[0,0]
)
axes[0,0].tick_params(axis='x', rotation=90)
axes[0,0].set_title("Top Winners")

# 2. Season vs Team1 Runs
sns.barplot(
    x='season',
    y='team1_runs',
    data=df,
    ax=axes[0,1]
)
axes[0,1].set_title("Season vs Team1 Runs")

# 3. Correlation Heatmap
corr = df.corr(numeric_only=True)

sns.heatmap(
    corr,
    annot=True,
    ax=axes[1,0]
)
axes[1,0].set_title("Correlation Heatmap")

# 4. Scatter Plot
sns.scatterplot(
    x='team1_runs',
    y='team2_runs',
    hue='winner',
    data=df,
    ax=axes[1,1]
)
axes[1,1].set_title("Team1 Runs vs Team2 Runs")

# 5. Histogram
sns.histplot(
    df['team1_runs'],
    kde=True,
    ax=axes[2,0]
)
axes[2,0].set_title("Distribution of Team1 Runs")

# 6. Box Plot
sns.boxplot(
    y='team1_runs',
    data=df,
    ax=axes[2,1]
)
axes[2,1].set_title("Box Plot of Team1 Runs")

plt.tight_layout()
plt.show()
```
