```python
import matplotlib.pyplot as plt
ipl_df.std(numeric_only=True)

batting_avg = ipl_df['batting_avg'].std()
ipl_df['batting_avg'].plot(kind='hist',bins=15,edgecolor='black',alpha=0.7,density=True)
ipl_df['batting_avg'].plot(kind='kde',color='red',ax=plt.gca(),secondary_y=False)
plt.title("Distribution of batting")
plt.xlabel("Batting average")
plt.ylabel("Density")
plt.xticks(rotation=90,ha='right')
plt.tight_layout()
plt.show()
```

# Create new Feature

```python
if{'runs','matches'}.issubset(ipl_df.columns):
  ipl_df['Runs_Per_Match']=ipl_df['runs']/ipl_df['matches']
```

# Value Count

```python
if 'Player_Type' not in ipl_df.columns and 'wickets' in ipl_df.columns:
  ipl_df['Player_Type'] = np.where(ipl_df['wickets']>50, 'Bowler','Batter')
ipl_df["Player_Type"].value_counts() if 'Player_Type' in ipl_df.columns else 'column unavailable'
```

# Group By Opearation

```python
ipl_df.groupby('Player_Type')[['runs','wickets']].mean()
```
