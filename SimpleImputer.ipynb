```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
```

```python
df = pd.read_csv('class_ipl_22-26_dataset_with_nulls.csv')
print(df.shape)
df.head()
```

```python
# missing value summery
df.isnull().sum()
```

```python
mean_df = df.copy()
mean_value = mean_df['team1_runs'].mean()
print(mean_value)
```

```python
mean_df['team1_runs']=mean_df['team1_runs'].fillna(mean_value)
mean_df['team1_runs'].isnull().sum()
```

```python
mean_df.isnull().sum()
```

```python
median_df = df.copy()
median_value = median_df['team2_runs'].median()
print(median_value)
```

```python
median_df['team2_runs']=median_df['team2_runs'].fillna(median_value)
median_df['team2_runs'].isnull().sum()
```

```python
mode_df = df.copy()
mode_df[['winner','player_of_match']].isnull().sum()
```

```python
mode_df['winner']= mode_df['winner'].fillna(mode_df['winner'].mode()[0])
mode_df['player_of_match']=mode_df['player_of_match'].fillna(mode_df['player_of_match'].mode()[0])

mode_df[['winner','player_of_match']].isnull().sum()
```

# Simple Imputer

```python
df.isnull().sum()
```

```python
mean_imp = SimpleImputer(strategy='mean')
median_imp = SimpleImputer(strategy='median')
mode_imp = SimpleImputer(strategy='most_frequent')

df['team1_wickets'] = mean_imp.fit_transform(df[['team1_wickets']])
df['team2_runs']=median_imp.fit_transform(df[['team2_runs']])
df[['winner']]=mode_imp.fit_transform(df[['winner']])
df[['player_of_match']]=mode_imp.fit_transform(df[['player_of_match']])

df.isnull().sum()
```
