# IPL Matplotlib Visualization

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('class_ipl_22-26_dataset.csv')
df.head()
```

```python
df.info()
```

```python
df['team1_wickets'].max()
```

```python
print(df['team2_wickets'].idxmax())
```

# Line Plot


```python
plt.plot(df['team1_runs'].tail(74))
plt.xlabel('Match Index')
plt.ylabel('Team 1 Runs')
plt.title('Team 1 Runs trends')
plt.show()
```

# Team Run Trend

```python
plt.plot(df['team2_runs'].tail(74))
plt.xlabel('Match Index')
plt.ylabel('Team 2 Runs')
plt.title('Team 2 Runs trends')
plt.show()
```

# Multiline Plot

```python
plt.figure(figsize=(12,6))
plt.plot(df['team1_runs'].tail(74),label='Team Runs',color='purple')
plt.plot(df['team1_wickets'].tail(74),label='Team Wickets',color='orange')
plt.xlabel('Runs')
plt.ylabel('Wickets')
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(12,6))
plt.plot(df['team1_runs'].tail(74),label='Team 1 Runs',color='green')
plt.plot(df['team2_runs'].tail(74),label='Team 2 Runs',color='maroon')
plt.xlabel('Runs')
plt.ylabel('Team1 Vs Team2 Runs')
plt.legend()
plt.grid(True)
plt.show()
```

# Scatter Plot

```python
match_indices = df.index[-74:] # match Numbers

plt.scatter(match_indices,df['team1_runs'].tail(74),c='blue',label='Team 1 Runs')
plt.scatter(match_indices,df['team2_runs'].tail(74),c='red',label='Team 2 Runs')
plt.xlabel('Match Index')
plt.ylabel('Runs')
plt.title('Team 1 Runs and Team 2 Runs per Match (last 74 Matches)')
plt.legend()
plt.show()
```

# Histogram


```python
plt.hist(df['team2_wickets'].tail(74),bins=10)
plt.xlabel('Wickets')
plt.ylabel('Number of Matches')
plt.title('Distribution Of Team 2 Wickets')
plt.show()
```

# Bar Chart

```python
toss_decisions = df['toss_decision'].value_counts()
plt.figure(figsize=(8,6))
plt.bar(toss_decisions.index,toss_decisions.values,color=['skyblue','lightcoral'])
plt.xlabel('Toss Decision')
plt.ylabel('Number of Matches')
plt.title('Distribution Of Toss Decisions')
plt.show()
```

```python
toss_decisions.index
```

```python
toss_decisions.values
```

```python
team_win = df['winner'].value_counts()

plt.figure(figsize=(8,6))
plt.bar(team_win.index,team_win.values,color=['skyblue','lightcoral'])
plt.xlabel('Wining Team')
plt.ylabel('Number Of Matches')
plt.title('Distribution Of Winning Teams')
plt.xticks(rotation=90)
plt.show()
```

# Box Plot

```python
plt.boxplot(df['team2_runs'])
plt.ylabel('Team 2 runs')
plt.title('Distribution Of Team 2 Runs')
plt.show()
```

```python
df['team2_runs'].describe()
```

# Pie Chart

```python
result_type_count = df['winner'].value_counts()

plt.figure(figsize=(8,8))
plt.pie(result_type_count.values,labels=result_type_count.index,pctdistance=0.5,radius=2,autopct='%1.1f%%',rotatelabels=True,startangle=90)
plt.title('Distribution Of Match Result Types')
plt.show()
```

```python
wicket_data = {
    '0-3':len(df[df['team1_wickets']<=3]),
    '4-7':len(df[(df['team1_wickets']>=4) & (df['team1_wickets']<=7)]),
    '8+':len(df[df['team1_wickets']>=8])
}
plt.pie(wicket_data.values(),labels=wicket_data.keys(),autopct='%1.1f%%')
plt.show()
```
