# Mini Case Study

```python
kohli_runs = np.array([45,72,33,61,88,54,40,76,92,58])
avg_runs = np.mean(kohli_runs)
print('Average:',avg_runs)
if avg_runs >50:
  print("Player is in Form")
else:
  print("Player is Not in Form")
```

```python
# Q1
kohli_runs = kohli_runs[-7:]
print(kohli_runs)
total_runs = np.sum(kohli_runs)
average_runs = np.mean(kohli_runs)
Highest_run = np.max(kohli_runs)
Lowest_run = np.min(kohli_runs)
print("Total Runs : ",total_runs)
print("Average Runs : ",average_runs)
print("Highest Runs : ",Highest_run)
print("Lowerst Runs : ",Lowest_run)
```

```python
# Q2
Virat = np.array([45,72,33,61,88,54,40,76,92,58])
Rohit = np.array([46,71,35,68,100,56,50,80,150,70])

virat_total = np.sum(Virat)
rohit_total = np.sum(Rohit)

Virat_avg = np.mean(Virat)
Rohit_avg = np.mean(Rohit)
diff = np.array([Virat>Rohit])
count=0

for i in diff:
  if i.any():
    count+=1

if(rohit_total>virat_total):
  print("Rohit Scored More Run")
  print("Runs:",rohit_total)
else:
  print("Virat Scored More Run")
  print("Runs:",virat_total)

print("Differnece in Average :",Rohit_avg-Virat_avg)

print("Number of Matches Virat Scored more runs than Rohit :",count)
```

```python
team_runs = np.array([[45,60,30],[50,70,40],[20,30,50],[60,40,80],[30,20,40]])

total =np.sum(team_runs,axis=1)
print("Total runs of each batter : ",np.sum(team_runs,axis=1))
print("Match wise team Total : ",np.sum(team_runs,axis=0))
print(np.max(total))
```

```python
arr = np.array([45,72,33,61,88,54,40,76,92,58])
mean = np.mean(arr)
std = np.std(arr)
var = np.var(arr)

print(mean)
print(std)
print(var)
```

```python
runs = np.array([1,4,0,6,2,1,4,0,1,6,2,1])
total_runs = np.sum(runs)
boundary = runs[runs>=4]
percentage = (np.sum(boundary)/total_runs)*100

print("Total Runs: ",total_runs)
print("Number of Boundaries: ",boundary.size)
print("Percentage: ",percentage)
```
