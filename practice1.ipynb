# Q15

Topic : Creating New Column

Question :
Create a new column that classifies matches based on the winning margin.

Explanation :
Creating new columns is useful for feature engineering

```python

conditions = [
    df['win_by_runs']<=10,
    (df['win_by_runs']>11) & (df['win_by_runs']<=40),
    df['win_by_runs']>=41
]
labels = ['Close Match', 'Comfortable Win', 'Dominant Win']
df1['margin']= np.select(conditions,labels,default="Unknown")
df1['margin']
```
