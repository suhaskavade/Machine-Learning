```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

ipl = pd.read_csv('/content/class_ipl_22-26_dataset.csv')
```

```python
ipl.shape
```

```python
ipl.columns
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    x='team1_wickets',
    y='team1_runs',
    data = ipl.tail(74)
)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(
    x='team1',
    y='team1_runs',
    hue='winner',
    data = ipl.tail(74)
).tick_params(axis="x",rotation=90)
plt.show()
```

```python
multi_var_cols = ['team1_runs', 'team2_runs', 'win_by_runs', 'win_by_wickets', 'season','result_type']
multi_var_df = ipl[multi_var_cols].head(100)

sns.pairplot(multi_var_df, hue='result_type', diag_kind='kde')

plt.suptitle('Multivarient Analysis: Runs, Wickets, season by result Type', y=1.02)
plt.show()
```

```python

numerical_cols = ['season', 'win_by_runs', 'win_by_wickets','team1_runs']

corr_df = ipl[numerical_cols]

corr_matrix = corr_df.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap = 'coolwarm',
    fmt=".2f"
)

plt.title('Correalation Heatmap of selected IPL MAtch data VAriables')
plt.show()
```
