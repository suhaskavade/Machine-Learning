```python
import numpy as np
import time

list1 = list(range(1000000))
start = time.time()

list1 = [ x+5 for x in list1]
end = time.time()
print("List Time : ",(end-start)*1000,"ms")

arr1 = np.array(range(1000000))
start = time.time()
arr1 = arr1+5
end = time.time()
print("array Time : ",(end-start)*1000,"ms")
```
